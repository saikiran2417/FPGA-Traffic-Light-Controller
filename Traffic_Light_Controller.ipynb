```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import time

# Create visual traffic light
red_light    = widgets.Button(description='🔴 RED',
                style={'button_color': 'gray'},
                layout=widgets.Layout(width='150px', height='80px'))

yellow_light = widgets.Button(description='🟡 YELLOW',
                style={'button_color': 'gray'},
                layout=widgets.Layout(width='150px', height='80px'))

green_light  = widgets.Button(description='🟢 GREEN',
                style={'button_color': 'gray'},
                layout=widgets.Layout(width='150px', height='80px'))

# Status labels
status_label = widgets.Label(value="🚦 Traffic Light Status: RUNNING")
timer_label  = widgets.Label(value="⏱️ Time in current state: 0s")
cycle_label  = widgets.Label(value="🔁 Total cycles completed: 0")

# Layout
dashboard = widgets.VBox([
    widgets.HTML("<h2>🚦 AXI Based Traffic Light Controller</h2>"),
    widgets.HTML("<b>Hardware: Xilinx Zynq 7020 | Board: PYNQ-Z2</b>"),
    widgets.HBox([red_light, yellow_light, green_light]),
    status_label,
    timer_label,
    cycle_label
])

display(dashboard)

# NEW FLOW
states    = ['RED', 'YELLOW', 'GREEN']   # ✅ Changed order
durations = {'RED': 2, 'YELLOW': 2, 'GREEN': 2}   # ✅ 1.5-2 sec

cycles = 0

print("\n▶️ Dashboard running... Press STOP (■) to stop kernel")

for _ in range(100):   # Run 100 cycles
    for state in states:

        # Update colors
        red_light.style.button_color    = 'red'    if state == 'RED'    else 'gray'
        yellow_light.style.button_color = 'orange' if state == 'YELLOW' else 'gray'
        green_light.style.button_color  = 'green'  if state == 'GREEN'  else 'gray'

        # Update status
        status_label.value = f"🚦 Current State: {state}"

        # Timer
        for t in range(durations[state], 0, -1):
            timer_label.value = f"⏱️ Time remaining: {t}s"
            time.sleep(1)

        # Count full cycle when GREEN ends
        if state == 'GREEN':
            cycles += 1
            cycle_label.value = f"🔁 Total cycles completed: {cycles}"

print("✅ Demo complete!")
```
